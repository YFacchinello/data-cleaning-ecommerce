```python
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
lucro_subcat = df.groupby("Sub-Category")["Profit"].sum().sort_values().reset_index()
plt.figure(figsize=(10, 6))

sns.barplot(data=lucro_subcat, x ="Profit", y="Sub-Category", palette="coolwarm")
plt.title("Lucro total por subcategoria", fontsize=14, fontweight="bold")
plt.xlabel("Lucro total(USD)", fontsize=12)
plt.ylabel("Subcategoria", fontsize=12)

plt.show()
```

```python
lucro_desconto_mesas = tables_df.groupby("Discount")["Profit"].sum().reset_index()
plt.figure(figsize=(9, 5))

ax = sns.barplot(data=lucro_desconto_mesas, x="Discount", y="Profit", palette="vlag")
plt.axhline(0, color="black", linewidth=1, linestyle='--')

plt.title("Impacto do desconto no lucro total das mesas (Tables)", fontsize=13, fontweight="bold")
plt.xlabel("Percentual de desconto concedido", fontsize=11)
plt.ylabel("Lucro total (USD)", fontsize=11)

plt.show()
```
